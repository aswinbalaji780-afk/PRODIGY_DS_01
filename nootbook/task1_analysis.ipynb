```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os

csv_path = os.path.join('..', 'data', 'train.csv')
df = pd.read_csv(csv_path)
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df['Age'], bins=20, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

# Age Distribution Insights

- Most passengers were between 20–40 years old.
- The distribution appears slightly right-skewed.
- Fewer elderly passengers were present in the dataset.

```python
plt.figure(figsize=(8,5))

sns.countplot(x='Sex', data=df)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

# Gender Distribution Insights

- Male passengers significantly outnumbered female passengers.
- The dataset shows a gender imbalance among passengers.
